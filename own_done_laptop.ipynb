```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder, LabelEncoder, StandardScaler, MinMaxScaler, RobustScaler, PolynomialFeatures
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, root_mean_squared_error, r2_score
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor
import joblib
```

```python
x = df.drop(columns=["Price"])
y = df["Price"]
```

```python
x.shape
```

```python
y.shape
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=22)
```

```python
cat_train = x_train[["Brand"]]
cat_test = x_test[["Brand"]]
```

```python
cat_train.shape
```

```python
cat_test.shape
```

```python
o = OneHotEncoder(sparse_output=False,handle_unknown="ignore")
cat_x_train = o.fit_transform(cat_train)
cat_x_train = pd.DataFrame(cat_x_train,columns=o.get_feature_names_out(["Brand"]),index=cat_train.index)
```

```python
cat_x_train.head(3)
```

```python
x_train.head(3)
```

```python
x_train_new = x_train.drop(columns=["Brand"])
x_train_new = x_train_new.join(cat_x_train)
```

```python
x_train_new.head(3)
```

```python
cat_x_test = o.transform(cat_test)
cat_x_test = pd.DataFrame(cat_x_test,columns=o.get_feature_names_out(["Brand"]),index=cat_test.index)
```

```python
cat_x_test.head(3)
```

```python
x_test.head(3)
```

```python
x_test_new = x_test.drop(columns=["Brand"])
x_test_new = x_test_new.join(cat_x_test)
```

```python
x_test_new.head(3)
```

```python
lr = LinearRegression()
lr.fit(x_train_new,y_train)
```

```python
lrypt = lr.predict(x_test_new)
```

```python
r = Ridge(alpha=2.0)
r.fit(x_train_new,y_train)
```

```python
rypt = r.predict(x_test_new)
```

```python
l = Lasso(alpha=0.5)
l.fit(x_train_new,y_train)
```

```python
lypt = l.predict(x_test_new)
```

```python
p = PolynomialFeatures(degree=2)
x_train_poly = p.fit_transform(x_train_new)
x_test_poly = p.transform(x_test_new)

pr = LinearRegression()
pr.fit(x_train_poly,y_train)
```

```python
prypt = pr.predict(x_test_poly)
```

```python
dt = DecisionTreeRegressor(random_state=22)
dt.fit(x_train_new,y_train)
```

```python
dtypt = dt.predict(x_test_new)
```

```python
sc = StandardScaler()
x_train_scaled = sc.fit_transform(x_train_new)
x_test_scaled = sc.transform(x_test_new)

kn = KNeighborsRegressor(n_neighbors=5)
kn.fit(x_train_scaled,y_train)
```

```python
knypt = kn.predict(x_test_scaled)
```

```python
rf = RandomForestRegressor(n_estimators=100,random_state=22)
rf.fit(x_train_new,y_train)
```

```python
rfypt = rf.predict(x_test_new)
```

```python
print(f"LinearRegression mse: {mean_squared_error(y_test,lrypt)}")
print(f"LinearRegression rmse: {root_mean_squared_error(y_test,lrypt)}")
print(f"LinearRegression r2: {r2_score(y_test,lrypt)}")
```

```python
print(f"Ridge mse: {mean_squared_error(y_test,rypt)}")
print(f"Ridge rmse: {root_mean_squared_error(y_test,rypt)}")
print(f"Ridge r2: {r2_score(y_test,rypt)}")
```

```python
print(f"Lasso pyt: {mean_squared_error(y_test,lypt)}")
print(f"Lasso pyt: {root_mean_squared_error(y_test,lypt)}")
print(f"Lasso pyt: {r2_score(y_test,lypt)}")
```

```python
print(f"PolynomialFeatures mse: {mean_squared_error(y_test,prypt)}")
print(f"PolynomialFeatures rmse: {root_mean_squared_error(y_test,prypt)}")
print(f"PolynomialFeatures r2: {r2_score(y_test,prypt)}")
```

```python
print(f"DecisionTreeRegressor mse: {mean_squared_error(y_test,dtypt)}")
print(f"DecisionTreeRegressor rmse: {root_mean_squared_error(y_test,dtypt)}")
print(f"DecisionTreeRegressor r2: {r2_score(y_test,dtypt)}")
```

```python
print(f"KNeighborsRegressor mse: {mean_squared_error(y_test,knypt)}")
print(f"KNeighborsRegressor rmse: {root_mean_squared_error(y_test,knypt)}")
print(f"KNeighborsRegressor r2: {r2_score(y_test,knypt)}")
```

```python
print(f"RandomForestRegressor mse: {mean_squared_error(y_test,rfypt)}")
print(f"RandomForestRegressor rmse: {root_mean_squared_error(y_test,rfypt)}")
print(f"RandomForestRegressor r2: {r2_score(y_test,rfypt)}")
```
